# Decision Tree Hyperparameter Optimization

```
# Import pandas for data handling
import pandas as pd  # used for tables (DataFrame)

# Import numpy for numerical operations
import numpy as np  # used for arrays and math

# Import matplotlib for plotting
import matplotlib.pyplot as plt  # used for graphs

# Import Decision Tree model
from sklearn.tree import DecisionTreeClassifier  # classification model

# Import train-test split
from sklearn.model_selection import train_test_split  # split data

# Import evaluation metrics
from sklearn.metrics import precision_score, recall_score, f1_score  # performance metrics
```

## Load dataset (replace with your own data)

```
# Example dataset (REMOVE and replace with your dataset)
from sklearn.datasets import make_classification  # function to create fake data

# Generate synthetic dataset
X, y = make_classification(
    n_samples=5000,        # number of samples
    n_features=10,         # number of features
    weights=[0.8, 0.2],    # imbalance (80% class 0, 20% class 1)
    random_state=42        # reproducibility
)

# Split dataset into training and testing sets
train_X, test_X, train_y, test_y = train_test_split(
    X,                     # features
    y,                     # target
    test_size=0.3,         # 30% for testing
    random_state=42        # reproducibility
)
```

## Define hyperparameters

```
# Define possible values for max_depth (tree depth)
max_depth_values = [3, 5, 10, 15, 20]  # different tree complexities

# Define possible values for min_samples_leaf (minimum samples per leaf)
min_samples_leaf_values = [1, 5, 10, 20, 50]  # control overfitting
```

## Baseline model

```
# Create baseline model (no tuning)
dtc3 = DecisionTreeClassifier(random_state=42)  # default tree

# Train baseline model
dtc3.fit(train_X, train_y)  # learn patterns from training data

# Predict on test data
y_pred_base = dtc3.predict(test_X)  # predictions

# Store baseline performance
baseline_results = {
    "model": "baseline",  # label
    "max_depth": None,  # no tuning
    "min_samples_leaf": None,  # no tuning
    "precision": precision_score(test_y, y_pred_base),  # precision metric
    "recall": recall_score(test_y, y_pred_base),  # recall metric
    "f1": f1_score(test_y, y_pred_base)  # f1 score
}

# Show baseline results
baseline_results
```

## Hyperparameter tuning

```
# Initialize list to store results
results = []  # empty list

# Loop over max_depth values
for md in max_depth_values:  # iterate depth values

    # Loop over min_samples_leaf values
    for msl in min_samples_leaf_values:  # iterate leaf values

        # Create model with current parameters
        model = DecisionTreeClassifier(
            max_depth=md,  # set depth
            min_samples_leaf=msl,  # set leaf size
            random_state=42  # reproducibility
        )

        # Train model
        model.fit(train_X, train_y)  # fit model

        # Predict on test data
        y_pred = model.predict(test_X)  # predictions

        # Store results
        results.append({
            "model": "dt",  # model type
            "max_depth": md,  # depth value
            "min_samples_leaf": msl,  # leaf size
            "precision": precision_score(test_y, y_pred),  # precision
            "recall": recall_score(test_y, y_pred),  # recall
            "f1": f1_score(test_y, y_pred)  # f1 score
        })
```

## Results analysis

```
# Add baseline to results
results.append(baseline_results)  # include baseline

# Convert list to DataFrame
results_df = pd.DataFrame(results)  # table format

# Show first rows
results_df.head()
```

## Top models

```
# Sort models by recall (descending)
top10 = results_df.sort_values("recall", ascending=False).head(10)  # best models

# Display top 10
top10
```

## Plot results

```
# Create labels for plot
labels = [
    f"d={row['max_depth']}, l={row['min_samples_leaf']}"  # label text
    for _, row in top10.iterrows()  # iterate rows
]

# Create plot
plt.figure()  # new figure

# Horizontal bar plot
plt.barh(labels, top10["recall"])  # plot recall

# Label x-axis
plt.xlabel("Recall")  # x label

# Add title
plt.title("Top Models by Recall")  # title

# Invert y-axis (best on top)
plt.gca().invert_yaxis()  # flip axis

# Show plot
plt.show()  # display graph
```
