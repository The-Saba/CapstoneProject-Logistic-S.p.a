```python
import pandas as pd
import numpy as np
import os

from utils.italian_provinces import italian_provinces
```

```python
BASE_DIR = os.getcwd()
file_path_csv = os.path.join(BASE_DIR, "dataset", "dataset.csv")
file_path_xlsx = os.path.join(BASE_DIR, "dataset", "Corrispondenza-Ateco-2022-vs-NACE-Rev.-2.xlsx")

df_csv = pd.read_csv(file_path_csv, sep=";", encoding="latin-1")
df_xlsx = pd.read_excel(file_path_xlsx, sheet_name="NACE Rev. 2 vs Ateco 2022")
```

```python
df_csv.rename(columns={
    "Natura": "Nature",
    "Fatturato": "Revenue",
    "Dipendenti": "Employees",
    "Provincia": "Province"
}, inplace=True)
```

```python
df_csv.loc[df_csv["Province"].notna(), ["City", "Region"]] = df_csv.loc[df_csv["Province"].notna(), "Province"].apply(
    lambda prov: pd.Series(italian_provinces.get(prov.strip().upper(), {"City": None, "Region": None}))
)
```

```python
cols = list(df_csv.columns)
province_idx = cols.index("Province")
# rimuovo se esistono già
for col in ["City", "Region"]:
    if col in cols:
        cols.remove(col)
# inserisco
cols = cols[:province_idx+1] + ["City", "Region"] + cols[province_idx+1:]
df_csv = df_csv[cols]
```

```python
# righe con Province presente ma City o Region mancanti
missing_city_region = df_csv[
    df_csv["Province"].notna() &
    (df_csv["City"].isna() | df_csv["Region"].isna())
]

print(f"Numero di righe non collegate: {len(missing_city_region)}")
missing_city_region[["Province", "City", "Region"]].head(20)  # mostra le prime 20
```

```python
df_csv = df_csv.dropna()

cols = [f"M-{i}" for i in range(36)]

df_csv = df_csv[df_csv[cols].sum(axis=1) != 0]
```

```python
df_csv['ATECO_norm'] = df_csv['ATECO'].str[2:]
df_csv['ATECO_Desc'] = (
    df_csv['ATECO_Desc']
    .str.lower()
    .str.replace(r'[^\w\s]', '', regex=True)
    .str.strip()
)
```

```python
df_xlsx.columns = ['NACE', 'NACE_Desc', 'ATECO_2007', 'ATECO_2007_Desc']
df_xlsx['ATECO_2007_Desc'] = (
    df_xlsx['ATECO_2007_Desc']
    .str.lower()
    .str.replace(r'[^\w\s]', '', regex=True)
    .str.strip()
)
```

```python
# 1. Primo tentativo: Join su ATECO_norm
df = df_csv.merge(
    df_xlsx,
    how="left",
    left_on="ATECO_norm",
    right_on="ATECO_2007"
)

df_2 = df_csv.merge(
    df_xlsx,
    how="left",
    left_on="ATECO_Desc",
    right_on="ATECO_2007_Desc",
    suffixes=('', '_backup')
)

df['NACE'] = df['NACE'].fillna(df_2['NACE'])
df['NACE_Desc'] = df['NACE_Desc'].fillna(df_2['NACE_Desc'])
df['ATECO_2007'] = df['ATECO_2007'].fillna(df_2['ATECO_2007'])
df['ATECO_2007_Desc'] = df['ATECO_2007_Desc'].fillna(df_2['ATECO_2007_Desc'])
```

```python
# righe con Province presente ma City o Region mancanti
missing_NACE = df[
    df["NACE"].isna()
]

print(f"Numero di righe non collegate: {len(missing_NACE)}")
missing_NACE[["ATECO", "ATECO_Desc"]].head(20)  # mostra le prime 20
```

```python
cols_to_drop = ['ATECO', 'ATECO_Desc', 'ATECO_2007', 'ATECO_2007_Desc', 'ATECO_norm', 'Nature']

df = df.drop(columns=cols_to_drop)

cols = df.columns.tolist()
last_two = cols[-2:]
rest = cols[:-2]
new_order = last_two + rest
df = df[new_order]
```
